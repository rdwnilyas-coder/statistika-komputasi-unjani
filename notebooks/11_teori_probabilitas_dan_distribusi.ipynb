````python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 5)

# Memuat dataset simulasi probabilitas
df_prob = pd.read_csv("../datasets/09_probability_distributions.csv")
print("Dataset Probabilitas dimuat. Total baris:", len(df_prob))
display(df_prob.head())
````

## 🎲 3. Teorema Bayes dalam Klasifikasi Probabilistik (Contoh: Spam Filter)


````python
# Implementasi Teorema Bayes: P(Spam | Kata 'Promo')
# Diketahui data historis:
p_spam = 0.20           # P(Spam) -> Prior
p_ham = 0.80            # P(Bukan Spam / Ham)
p_promo_given_spam = 0.85 # P(Kata 'Promo' | Spam) -> Likelihood
p_promo_given_ham = 0.05  # P(Kata 'Promo' | Ham)

# Total Evidence: P(Kata 'Promo')
p_promo = (p_promo_given_spam * p_spam) + (p_promo_given_ham * p_ham)

# Posterior: P(Spam | Kata 'Promo')
p_spam_given_promo = (p_promo_given_spam * p_spam) / p_promo

print("=== Perhitungan Teorema Bayes ===")
print(f"P(Prior Spam)                  : {p_spam*100:.1f}%")
print(f"P(Promo | Spam) (Likelihood)   : {p_promo_given_spam*100:.1f}%")
print(f"P(Promo) (Evidence)            : {p_promo*100:.1f}%")
print(f">> P(Spam | Promo) (Posterior) : {p_spam_given_promo*100:.2f}%")
````

## 📊 4. Pemodelan Distribusi Diskrit (Binomial & Poisson)


````python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 1. Distribusi Binomial: Sukses k dari n=10 percobaan (p=0.5)
n_trials, p_succ = 10, 0.5
k_vals = np.arange(0, 11)
binomial_pmf = stats.binom.pmf(k_vals, n_trials, p_succ)

axes[0].bar(k_vals, binomial_pmf, color='navy', alpha=0.8, edgecolor='black')
axes[0].set_title('Distribusi Peluang Binomial (n=10, p=0.5)', fontweight='bold')
axes[0].set_xlabel('Jumlah Sukses (k)')
axes[0].set_ylabel('Probabilitas P(X = k)')
axes[0].set_xticks(k_vals)

# 2. Distribusi Poisson: Kedatangan request server (lambda=12 req/menit)
lambda_rate = 12
k_poisson = np.arange(0, 25)
poisson_pmf = stats.poisson.pmf(k_poisson, lambda_rate)

axes[1].bar(k_poisson, poisson_pmf, color='coral', alpha=0.8, edgecolor='black')
axes[1].axvline(lambda_rate, color='red', linestyle='--', label=f'Mean Rate λ={lambda_rate}')
axes[1].set_title('Distribusi Peluang Poisson (λ = 12 request/menit)', fontweight='bold')
axes[1].set_xlabel('Jumlah Request Datang (k)')
axes[1].set_ylabel('Probabilitas P(X = k)')
axes[1].legend()

plt.tight_layout()
plt.show()
````

## 📈 5. Distribusi Kontinu Normal (Gaussian) & Kurva Empiris Z-Score


````python
mu, sigma = 100, 15
x_axis = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
pdf_normal = stats.norm.pdf(x_axis, mu, sigma)

plt.figure(figsize=(10, 5))
plt.plot(x_axis, pdf_normal, color='navy', lw=2.5, label='Kurva Distribusi Normal N(100, 15²)')

# Shading Aturan Empiris 68% (mu +/- 1 sigma)
plt.fill_between(x_axis, pdf_normal, where=((x_axis >= mu-sigma) & (x_axis <= mu+sigma)), 
                 color='teal', alpha=0.3, label='68.2% Data (μ ± 1σ)')
# Shading Aturan Empiris 95% (mu +/- 2 sigma)
plt.fill_between(x_axis, pdf_normal, where=((x_axis >= mu-2*sigma) & (x_axis <= mu-sigma)) | ((x_axis >= mu+sigma) & (x_axis <= mu+2*sigma)), 
                 color='coral', alpha=0.3, label='95.4% Data (μ ± 2σ)')

plt.axvline(mu, color='red', linestyle='--', label='Mean μ = 100')
plt.title('Kurva Distribusi Normal & Kaidah Empiris Probabilitas', fontweight='bold')
plt.xlabel('Nilai Variabel (X)')
plt.ylabel('Densitas Probabilitas f(x)')
plt.legend()
plt.show()

# Probabilitas menghitung skor > 120
prob_greater_120 = 1 - stats.norm.cdf(120, mu, sigma)
print(f"Probabilitas X > 120 (Z = {(120-mu)/sigma:.2f}): {prob_greater_120*100:.2f}%")
````

### Q&A
* **Apa signifikansi Teorema Bayes dalam machine learning?** Teorema Bayes memungkinkan sistem memperbarui probabilitas hipotesis saat bukti (*evidence/data baru*) bertambah, menjadi fondasi algoritma Naive Bayes Classifier dan Bayesian Inference.
* **Kapan menggunakan Poisson vs Binomial?** Binomial digunakan untuk jumlah percobaan $n$ yang tetap dengan peluang sukses $p$. Poisson digunakan saat $n$ sangat besar dan $p$ kecil, yaitu untuk menghitung laju kejadian per satuan waktu/ruang (misal: trafik request server).

### Data Analysis Key Findings
* Penerapan Teorema Bayes membuktikan bahwa kemunculan kata 'Promo' melipatgandakan peluang email adalah Spam dari Prior **20%** menjadi Posterior **80.95%**.
* Distribusi Normal membagi peluang secara simetris di mana **95.4%** data terkonsentrasi di dalam rentang $\mu \pm 2\sigma$.
