## 📖 1. Teori Probabilitas dan Distribusi Peluang

Teori Probabilitas adalah pilar matematis paling fundamental dalam pembelajaran mesin (*Machine Learning*), inferensi statistik, dan pemodelan ketidakpastian sistem komputasi:
1. **Peluang Bersyarat & Teorema Bayes**:
   - Peluang terjadinya peristiwa $A$ dengan syarat peristiwa $B$ telah terjadi: $P(A|B) = 
rac{P(A \cap B)}{P(B)}$.
   - **Teorema Bayes**: Mengintegrasikan informasi awal (*Prior Probability*) dengan data bukti baru (*Likelihood*) untuk menghasilkan peluang mutakhir (*Posterior Probability*):
     $$P(A|B) = 
rac{P(B|A) \cdot P(A)}{P(B)}$$
2. **Distribusi Peluang Diskrit**:
   - **Distribusi Binomial**: Memodelkan jumlah $k$ keberhasilan dari $n$ percobaan independen biner (sukses/gagal) dengan probabilitas sukses $p$.
   - **Distribusi Poisson**: Memodelkan jumlah kejadian diskrit dalam interval waktu atau ruang tertentu dengan laju rata-rata $\lambda$ (misal: trafik request masuk ke server web per menit).
3. **Distribusi Peluang Kontinu (Normal / Gaussian)**:
   - Distribusi berbentuk kurva lonceng simetris $N(\mu, \sigma^2)$ dengan Kaidah Empiris:
     * $68.26\%$ data berada di dalam $\mu \pm 1\sigma$.
     * $95.44\%$ data berada di dalam $\mu \pm 2\sigma$.
     * $99.73\%$ data berada di dalam $\mu \pm 3\sigma$.
   - **Standard Normal Z-Score**: $Z = 
rac{X - \mu}{\sigma} \sim N(0, 1)$.

## 💻 4. Eksekusi Komputasi Python: Teorema Bayes & Distribusi Peluang


```python
# 1. Implementasi Teorema Bayes (Klasifikasi Email Spam)
p_spam = 0.20             # P(Spam) -> Prior
p_ham = 0.80              # P(Ham / Bukan Spam)
p_promo_given_spam = 0.85 # P('Promo' | Spam) -> Likelihood
p_promo_given_ham = 0.05  # P('Promo' | Ham)

# Total Evidence: P('Promo')
p_promo = (p_promo_given_spam * p_spam) + (p_promo_given_ham * p_ham)

# Posterior: P(Spam | 'Promo')
p_spam_given_promo = (p_promo_given_spam * p_spam) / p_promo

print("=== Hasil Inferensi Teorema Bayes ===")
print(f"Peluang Prior Spam           : {p_spam*100:.1f}%")
print(f"Peluang Evidence Kata 'Promo': {p_promo*100:.1f}%")
print(f">> Peluang Posterior P(Spam|Promo) : {p_spam_given_promo*100:.2f}%")
```

```python
# 2. Visualisasi Distribusi Poisson & Normal Gaussian
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Subplot 1: Distribusi Poisson Kedatangan Request Server (lambda=12 req/min)
lambda_val = 12
k_range = np.arange(0, 25)
poisson_pmf = stats.poisson.pmf(k_range, lambda_val)

axes[0].bar(k_range, poisson_pmf, color='#1A365D', alpha=0.85, edgecolor='black')
axes[0].axvline(lambda_val, color='#EA580C', linestyle='--', lw=2, label=f'Mean Rate (λ = {lambda_val})')
axes[0].set_title('Distribusi Poisson Laju Request Server (λ = 12)', fontweight='bold')
axes[0].set_xlabel('Jumlah Request Masuk per Menit (k)')
axes[0].set_ylabel('Probabilitas P(X = k)')
axes[0].legend()

# Subplot 2: Distribusi Normal & Kaidah Empiris
x_norm = np.linspace(-3.5, 3.5, 200)
pdf_norm = stats.norm.pdf(x_norm, 0, 1)
axes[1].plot(x_norm, pdf_norm, color='#1A365D', lw=2.5)
axes[1].fill_between(x_norm, pdf_norm, where=((x_norm >= -1) & (x_norm <= 1)), color='#EA580C', alpha=0.35, label='68.2% (μ ± 1σ)')
axes[1].fill_between(x_norm, pdf_norm, where=((x_norm >= -2) & (x_norm <= -1)) | ((x_norm >= 1) & (x_norm <= 2)), color='#2B6CB0', alpha=0.25, label='95.4% (μ ± 2σ)')
axes[1].set_title('Kaidah Empiris Distribusi Normal Standar N(0, 1)', fontweight='bold')
axes[1].set_xlabel('Nilai Z-Score')
axes[1].set_ylabel('Kerapatan Peluang f(z)')
axes[1].legend()

plt.tight_layout()
plt.show()
```
