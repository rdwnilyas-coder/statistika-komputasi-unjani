## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `11_software_metrics_sdlc.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

# 1. Ingestion Data Metrik SDLC
df_sdlc = pd.read_csv("../datasets/11_software_metrics_sdlc.csv")
print("Dimensi Metrik SDLC:", df_sdlc.shape)
display(df_sdlc.head())

# 2. Analisis Statistik Deskriptif & Kerapatan Cacat (Defect Density)
summary_metrics = df_sdlc.groupby('team')[['lines_of_code', 'cyclomatic_complexity', 'defect_density_kloc']].mean()
print("=== Rata-rata Metrik Software per Tim Pengembang ===")
display(summary_metrics)
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Hubungan Kompleksitas Siklomatis terhadap Defect Density
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_sdlc, x='cyclomatic_complexity', y='defect_density_kloc', hue='team', s=100, palette='viridis')
plt.title("Hubungan Kompleksitas Kode vs Kerapatan Cacat (Defect Density per KLOC)", fontsize=13, fontweight='bold')
plt.xlabel("Cyclomatic Complexity")
plt.ylabel("Defect Density (Cacat / 1000 Baris)")
plt.tight_layout()
plt.show()
```
