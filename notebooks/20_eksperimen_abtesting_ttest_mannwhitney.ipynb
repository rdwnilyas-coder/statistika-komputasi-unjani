## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `16_refactoring_latency_ttest.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
from scipy import stats

# 1. Ingestion Data Latensi Sebelum & Sesudah Refactoring
df_refactor = pd.read_csv("../datasets/16_refactoring_latency_ttest.csv")
display(df_refactor.head())

# 2. Uji Beda Berpasangan (Paired Samples t-Test)
t_stat, p_val = stats.ttest_rel(df_refactor['latency_before_ms'], df_refactor['latency_after_ms'])
print(f"=== Hasil Paired t-Test (Latensi API) ===")
print(f"Nilai t-hitung: {t_stat:.4f}")
print(f"p-value: {p_val:.4e}")
print("Kesimpulan:", "PERBAIKAN SIGNIFIKAN (p < 0.05)" if p_val < 0.05 else "TIDAK ADA PERBEDAAN SIGNIFIKAN")
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Penurunan Latensi Sebelum vs Sesudah Refactoring
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
df_melt = pd.melt(df_refactor, value_vars=['latency_before_ms', 'latency_after_ms'], var_name='Tahapan', value_name='Latensi_ms')
sns.boxplot(data=df_melt, x='Tahapan', y='Latensi_ms', palette=['#EA580C', '#0D9488'])
plt.title("Komparasi Latensi Endpoint API Sebelum vs Sesudah Refactoring (ms)", fontsize=13, fontweight='bold')
plt.xlabel("Kondisi Sistem")
plt.ylabel("Latensi Respon (ms)")
plt.tight_layout()
plt.show()
```
