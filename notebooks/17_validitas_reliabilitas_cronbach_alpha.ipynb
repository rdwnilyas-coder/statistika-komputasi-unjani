## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `13_survey_reliability_items.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np

# 1. Ingestion Data Butir Kuesioner
# [Google Colab] Jika menjalankan di Colab, gunakan tautan direct raw GitHub berikut:
# url_13_survey_reliability_items = "https://raw.githubusercontent.com/rdwnilyas-coder/statistika-komputasi-unjani/refs/heads/main/datasets/13_survey_reliability_items.csv"
# df_items = pd.read_csv(url_13_survey_reliability_items)
df_items = pd.read_csv("../datasets/13_survey_reliability_items.csv")
item_cols = [c for c in df_items.columns if c.startswith('item_')]
df_scores = df_items[item_cols]

# 2. Perhitungan Cronbach's Alpha
k = df_scores.shape[1] # jumlah butir
item_variances = df_scores.var(axis=0, ddof=1).sum()
total_variance = df_scores.sum(axis=1).var(ddof=1)

cronbach_alpha = (k / (k - 1)) * (1 - (item_variances / total_variance))
print(f"Jumlah Butir Soal (k): {k}")
print(f"Koefisien Cronbach's Alpha: {cronbach_alpha:.4f}")
print("Status Reliabilitas:", "SANGAT RELIABEL (Alpha >= 0.70)" if cronbach_alpha >= 0.70 else "TIDAK RELIABEL")
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Korelasi Tiap Butir dengan Total Skor (Item-Total Correlation)
total_score = df_scores.sum(axis=1)
item_corrs = df_scores.apply(lambda col: col.corr(total_score))

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 4))
sns.barplot(x=item_corrs.index, y=item_corrs.values, palette='Blues_r')
plt.axhline(0.30, color='red', linestyle='--', label='Batas Validitas Minimum (r = 0.30)')
plt.title("Uji Validitas Butir Pertanyaan (Item-Total Correlation)", fontsize=13, fontweight='bold')
plt.xlabel("Kode Butir Kuesioner")
plt.ylabel("Korelasi Pearson (r)")
plt.legend()
plt.tight_layout()
plt.show()
```
