## 📖 1. Pengantar Teori & Metrik Rekayasa Perangkat Lunak

Ketika penilaian melibatkan data ordinal dengan penalti bertingkat (misal: selisih penilaian *Minor* ke *Critical* lebih fatal dibanding *Minor* ke *Major*), digunakan **Weighted Cohen's Kappa**. Selain itu, jika pengujian heuristik usabilitas melibatkan lebih dari dua evaluator ($k > 2$ ahli UI/UX), digunakan **Fleiss' Kappa** untuk menguji konsistensi penilaian antar banyak penilai.

## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `15_heuristic_evaluation_fleiss.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score

# 1. Ingestion Data Evaluasi Heuristik Multi-Pakar
df_fleiss = pd.read_csv("../datasets/15_heuristic_evaluation_fleiss.csv")
display(df_fleiss.head())

# 2. Perhitungan Quadratic Weighted Kappa Antar 2 Pakar
w_kappa = cohen_kappa_score(df_fleiss['expert_1'], df_fleiss['expert_2'], weights='quadratic')
print(f"Quadratic Weighted Kappa (Expert 1 vs Expert 2): {w_kappa:.4f}")
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Frekuensi Kesepakatan per Komponen UI
eval_cols = ['expert_1', 'expert_2', 'expert_3', 'expert_4']
pass_counts = (df_fleiss[eval_cols] == 'Pass').sum(axis=1)

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 4))
sns.histplot(pass_counts, discrete=True, color='#0F4C5C')
plt.title("Distribusi Jumlah Pakar yang Menilai 'Pass' per Komponen UI", fontsize=13, fontweight='bold')
plt.xlabel("Jumlah Pakar yang Sepakat (0 sd 4)")
plt.ylabel("Jumlah Komponen UI")
plt.tight_layout()
plt.show()
```
