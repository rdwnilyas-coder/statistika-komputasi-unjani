## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `18_software_reliability_failures.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np

# 1. Ingestion Data Keandalan Software
df_rel = pd.read_csv("../datasets/18_software_reliability_failures.csv")
display(df_rel.head())

# 2. Ringkasan Waktu Operasi dan Kejadian Kegagalan
print("=== Statistik Keandalan Sistem ===")
print(df_rel.groupby('architecture_version')[['runtime_hours', 'failure_event']].mean())
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Distribusi Waktu Berjalan (Runtime Hours)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.histplot(data=df_rel, x='runtime_hours', hue='architecture_version', kde=True, palette=['#0F4C5C', '#F59E0B'], bins=20)
plt.title("Distribusi Waktu Berjalan Instance Server Sebelum Kegagalan (Jam)", fontsize=13, fontweight='bold')
plt.xlabel("Runtime (Jam)")
plt.ylabel("Frekuensi Instance")
plt.tight_layout()
plt.show()
```
