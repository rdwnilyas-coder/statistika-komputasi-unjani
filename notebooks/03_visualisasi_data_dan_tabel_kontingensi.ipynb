````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv("../datasets/01_ecommerce_sales_eda.csv")
print("Dataset berhasil dimuat!")
````

## 📑 3. Pembuatan Tabel Kontingensi (Crosstab) dan Proporsi


````python
# Tabel Kontingensi Frekuensi Bersama
crosstab_raw = pd.crosstab(df['category'], df['payment_method'], margins=True, margins_name="Total")

print("=== Tabel Kontingensi (Frekuensi Absolut) ===")
display(crosstab_raw)

# Tabel Proporsi Normalisasi Berdasarkan Baris (Row Percentage)
crosstab_norm = pd.crosstab(df['category'], df['payment_method'], normalize='index') * 100

print("\n=== Tabel Proporsi Pembayaran per Kategori Produk (%) ===")
display(crosstab_norm.round(1))
````

## 🎨 4. Visualisasi Kontingensi: Heatmap & Grouped Bar Chart


````python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Subplot 1: Heatmap Tabel Kontingensi
sns.heatmap(crosstab_norm, annot=True, fmt=".1f", cmap="YlGnBu", ax=axes[0], cbar_kws={'label': 'Persentase (%)'})
axes[0].set_title('Heatmap Preferensi Metode Bayar per Kategori (%)', fontweight='bold')
axes[0].set_ylabel('Kategori Produk')
axes[0].set_xlabel('Metode Pembayaran')

# Subplot 2: Grouped Bar Chart
df_plot = df.groupby(['category', 'payment_method']).size().reset_index(name='count')
sns.barplot(data=df_plot, x='category', y='count', hue='payment_method', ax=axes[1], palette='tab10')
axes[1].set_title('Distribusi Metode Bayar Berdasarkan Kategori', fontweight='bold')
axes[1].set_xlabel('Kategori Produk')
axes[1].set_ylabel('Jumlah Transaksi')
axes[1].tick_params(axis='x', rotation=30)
axes[1].legend(title='Metode Bayar')

plt.tight_layout()
plt.show()
````

## 🎻 5. Visualisasi Multivariat: Violin Plot & Scatter Komparasi


````python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Subplot 1: Violin Plot Total Transaksi per Kategori
sns.violinplot(data=df[df['total_amount_k'] < 2000], x='category', y='total_amount_k', ax=axes[0], palette='Set2')
axes[0].set_title('Distribusi Nilai Transaksi per Kategori (Violin Plot)', fontweight='bold')
axes[0].set_xlabel('Kategori')
axes[0].set_ylabel('Total Belanja (k IDR)')
axes[0].tick_params(axis='x', rotation=30)

# Subplot 2: Scatter Plot Diskon vs. Rating dengan Hue Gender
sns.scatterplot(data=df, x='discount_pct', y='rating_score', hue='gender', size='quantity', ax=axes[1], palette='coolwarm', sizes=(40, 200), alpha=0.8)
axes[1].set_title('Hubungan Diskon (%) vs. Rating Skor', fontweight='bold')
axes[1].set_xlabel('Diskon (%)')
axes[1].set_ylabel('Rating Skor (1 - 5)')

plt.tight_layout()
plt.show()
````
