## 📖 1. Visualisasi Data Eksploratif dan Tabel Kontingensi

Dalam analisis data modern, visualisasi bukan sekadar representasi grafis, melainkan instrumen diagnostik primer untuk menemukan pola tersembunyi, korelasi non-linier, dan anomali struktural (*John Tukey's Exploratory Data Analysis*):
1. **Analisis Bivariat Kategorikal (Tabel Kontingensi / Crosstab)**:
   - Menghitung frekuensi gabungan $O_{ij}$ antara dua variabel kategorikal (misal: Metode Pembayaran vs Tingkat Kepuasan Pelanggan).
   - Memungkinkan perhitungan probabilitas marjinal, probabilitas bersyarat, dan menjadi basis uji independensi Chi-Square ($\chi^2$).
2. **Visualisasi Distribusi & Komparasi Antar Grup**:
   - **Violin Plot**: Menggabungkan ketelitian Boxplot dengan kurva estimasi densitas kernel (*Kernel Density Estimation* / KDE).
   - **Heatmap Kontingensi**: Memvisualisasikan intensitas frekuensi atau proporsi persentase antar sel kategori.

Studi kasus menganalisis hubungan antara metode pembayaran yang dipilih pelanggan dengan tingkat kepuasan layanan dan jumlah transaksi yang dibelanjakan.

**Tahapan Komputasi:**
1. Membangun tabel kontingensi 2 arah (*Two-Way Crosstab*) dengan proporsi normalisasi baris dan kolom.
2. Memvisualisasikan matriks kontingensi menggunakan Heatmap beranotasi.
3. Melakukan perbandingan multivariat nilai transaksi antar kelompok kategori menggunakan Violin Plot.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
# [Google Colab] Jika menjalankan di Colab, gunakan tautan direct raw GitHub berikut:
# url_01_ecommerce_sales_eda = "https://raw.githubusercontent.com/rdwnilyas-coder/statistika-komputasi-unjani/refs/heads/main/datasets/01_ecommerce_sales_eda.csv"
# df_sales = pd.read_csv(url_01_ecommerce_sales_eda)
df_sales = pd.read_csv("../datasets/01_ecommerce_sales_eda.csv")
```

## 💻 4. Eksekusi Komputasi Python & Tabulasi Silang


```python
# 1. Pembentukan Tabel Kontingensi (Crosstab)
ct_counts = pd.crosstab(df_sales['payment_method'], df_sales['customer_satisfaction'], margins=True, margins_name="Total")
ct_props = pd.crosstab(df_sales['payment_method'], df_sales['customer_satisfaction'], normalize='index') * 100

print("=== Tabel Kontingensi Frekuensi Observasi ===")
display(ct_counts)

print("=== Tabel Kontingensi Proporsi Persentase Baris (%) ===")
display(ct_props.round(1))
```

```python
# 2. Visualisasi Heatmap dan Violin Plot
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Heatmap Tabel Kontingensi
sns.heatmap(ct_props, annot=True, fmt=".1f", cmap="Blues", ax=axes[0], cbar_kws={'label': 'Persentase Baris (%)'})
axes[0].set_title('Heatmap Proporsi Kepuasan per Metode Pembayaran', fontweight='bold')
axes[0].set_xlabel('Tingkat Kepuasan')
axes[0].set_ylabel('Metode Pembayaran')

# Violin Plot Multivariat
sns.violinplot(data=df_sales, x='payment_method', y='total_amount_k', hue='customer_satisfaction', 
               palette='Set2', split=False, ax=axes[1])
axes[1].set_title('Distribusi Belanja per Metode Pembayaran & Kepuasan', fontweight='bold')
axes[1].set_xlabel('Metode Pembayaran')
axes[1].set_ylabel('Total Belanja (K IDR)')
axes[1].legend(title='Kepuasan', loc='upper right')

plt.tight_layout()
plt.show()
```
