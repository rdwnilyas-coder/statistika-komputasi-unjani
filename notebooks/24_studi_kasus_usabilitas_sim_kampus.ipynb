## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `20_case_sim_campus_usability.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
from scipy import stats

# 1. Ingestion Data Studi Kasus SIM Kampus
df_sim = pd.read_csv("../datasets/20_case_sim_campus_usability.csv")
display(df_sim.head())

# 2. Uji Komparasi Skor SUS SIM Lama vs SIM Redesign
old_sim = df_sim[df_sim['system_version'] == 'SIM_Lama']['sus_score']
new_sim = df_sim[df_sim['system_version'] == 'SIM_Redesign']['sus_score']

t_stat, p_val = stats.ttest_ind(new_sim, old_sim)
print(f"Rata-rata Skor SIM Lama: {old_sim.mean():.2f}")
print(f"Rata-rata Skor SIM Redesign: {new_sim.mean():.2f}")
print(f"Uji Independent t-Test: t = {t_stat:.4f}, p = {p_val:.4e}")
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Perbandingan Skor Usabilitas per Kelompok Pengguna
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.barplot(data=df_sim, x='user_group', y='sus_score', hue='system_version', palette=['#64748B', '#0F4C5C'])
plt.axhline(68.0, color='red', linestyle='--', label='Standar Kelayakan Rata-rata (68.0)')
plt.title("Komparasi Skor Usabilitas SIM Kampus Lama vs Redesign per Kelompok Pengguna", fontsize=13, fontweight='bold')
plt.xlabel("Kelompok Pengguna")
plt.ylabel("Rata-rata Skor SUS")
plt.legend()
plt.tight_layout()
plt.show()
```
