## 📖 1. Pengantar Teori & Metrik Rekayasa Perangkat Lunak

## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `12_usability_sus_tam_survey.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Ingestion Data Survei Usabilitas SUS & TAM
df_sus = pd.read_csv("../datasets/12_usability_sus_tam_survey.csv")
print("Dimensi Data Survei SUS:", df_sus.shape)
display(df_sus.head())

# 2. Verifikasi Perhitungan Skor SUS Standar Brooke (1986)
# Butir Ganjil (Positif): Skor - 1 | Butir Genap (Negatif): 5 - Skor
pos_items = ['sus_q1', 'sus_q3', 'sus_q5', 'sus_q7', 'sus_q9']
neg_items = ['sus_q2', 'sus_q4', 'sus_q6', 'sus_q8', 'sus_q10']

computed_sus = ((df_sus[pos_items].sum(axis=1) - 5) + (25 - df_sus[neg_items].sum(axis=1))) * 2.5
print(f"Rata-rata Skor SUS Keseluruhan: {computed_sus.mean():.2f} (Standar Industri: 68.0)")
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Distribusi Skor SUS berdasarkan Peran Pengguna
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_sus, x='role', y='calculated_sus_score', palette='Set2')
plt.axhline(68.0, color='red', linestyle='--', label='Ambang Batas Rata-rata Industri (68.0)')
plt.title("Distribusi Skor System Usability Scale (SUS) per Kelompok Pengguna", fontsize=13, fontweight='bold')
plt.xlabel("Peran Pengguna")
plt.ylabel("Skor SUS (0 - 100)")
plt.legend()
plt.tight_layout()
plt.show()
```
