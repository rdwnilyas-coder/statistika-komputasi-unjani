## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `19_user_churn_survival.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np

# 1. Ingestion Data Retensi Pengguna SaaS
df_churn = pd.read_csv("../datasets/19_user_churn_survival.csv")
display(df_churn.head())

# 2. Statistik Durasi Berlangganan (Tenure Months)
print("=== Rata-rata Masa Aktif per Paket Langganan ===")
display(df_churn.groupby('subscription_tier')[['tenure_months', 'churn_event', 'monthly_active_days']].mean())
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Durasi Masa Aktif Berdasarkan Paket Langganan
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.boxplot(data=df_churn, x='subscription_tier', y='tenure_months', hue='churn_event', palette=['#0D9488', '#EA580C'])
plt.title("Distribusi Masa Aktif Pengguna (Tenure Months) berdasarkan Paket & Status Churn", fontsize=13, fontweight='bold')
plt.xlabel("Paket Langganan")
plt.ylabel("Lama Berlangganan (Bulan)")
plt.legend(title='Status Churn (0=Aktif, 1=Berhenti)')
plt.tight_layout()
plt.show()
```
