## 📖 1. Pengantar Teori & Metrik Rekayasa Perangkat Lunak

Ketika mengevaluasi efektivitas antarmuka perangkat lunak pada berbagai faktor simultan (misal: 3 Tipe Perangkat $\times$ 2 Jenis Tata Letak Navigasi), uji-t tidak lagi memadai karena meningkatkan risiko *Family-Wise Error Rate*. **Two-Way ANOVA** digunakan untuk menguji Efek Utama (*Main Effects*) masing-masing faktor serta Efek Interaksi (*Interaction Effect*) antar faktor terhadap waktu penyelesaian tugas pengguna (*Task Completion Time*).

## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `17_ui_interaction_anova.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# 1. Ingestion Data Interaksi UI/UX
# [Google Colab] Jika menjalankan di Colab, gunakan tautan direct raw GitHub berikut:
# url_17_ui_interaction_anova = "https://raw.githubusercontent.com/rdwnilyas-coder/statistika-komputasi-unjani/refs/heads/main/datasets/17_ui_interaction_anova.csv"
# df_ui = pd.read_csv(url_17_ui_interaction_anova)
df_ui = pd.read_csv("../datasets/17_ui_interaction_anova.csv")
display(df_ui.head())

# 2. Pemodelan Two-Way ANOVA dengan Interaksi
model = ols('task_completion_time_sec ~ C(device_type) + C(navigation_layout) + C(device_type):C(navigation_layout)', data=df_ui).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print("=== TABEL TWO-WAY ANOVA INTERAKSI UI ===")
display(anova_table)
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Interaction Plot
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))
sns.pointplot(data=df_ui, x='device_type', y='task_completion_time_sec', hue='navigation_layout', markers=['o', 's'], linestyles=['-', '--'], capsize=0.1, palette=['#0F4C5C', '#F59E0B'])
plt.title("Interaction Plot: Pengaruh Tipe Perangkat & Navigasi terhadap Waktu Tugas", fontsize=13, fontweight='bold')
plt.xlabel("Tipe Perangkat")
plt.ylabel("Waktu Penyelesaian Tugas (detik)")
plt.tight_layout()
plt.show()
```

### 🎯 Kesimpulan & Actionable Insights

2. **Rekomendasi Desain**: Layout *Bottom Navigation* jauh lebih unggul pada perangkat Mobile, sedangkan *Drawer* lebih cocok untuk Desktop.
