## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `21_case_cloud_reliability_churn.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
import numpy as np

# 1. Ingestion Data Keandalan Cloud & Retensi Akun
df_cloud = pd.read_csv("../datasets/21_case_cloud_reliability_churn.csv")
display(df_cloud.head())

# 2. Analisis Pengaruh Pelanggaran SLA terhadap Tingkat Churn
print("=== Analisis Ketersediaan Layanan & Churn ===")
display(df_cloud.groupby('sla_tier')[['service_availability_pct', 'sla_breach_count', 'account_churn']].mean())
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Hubungan Ketersediaan Layanan vs Pelanggaran SLA
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_cloud, x='service_availability_pct', y='sla_breach_count', hue='account_churn', palette=['#0D9488', '#EA580C'], s=100)
plt.axvline(99.9, color='red', linestyle='--', label='Batas Minimum SLA (99.9%)')
plt.title("Korelasi Ketersediaan Layanan Cloud (%) vs Frekuensi Insiden & Churn Akun", fontsize=13, fontweight='bold')
plt.xlabel("Ketersediaan Layanan (%)")
plt.ylabel("Jumlah Pelanggaran SLA")
plt.legend(title='Status Churn')
plt.tight_layout()
plt.show()
```
