## 📖 1. Pengantar Teori & Metrik Rekayasa Perangkat Lunak

Dalam pengujian perangkat lunak (*Software Quality Assurance*), penentuan keparahan bug (*Bug Severity*) sering kali dinilai oleh dua rater/evaluator independen. Menghitung persentase kesepakatan (*Percentage Agreement*) sering kali menyesatkan karena tidak memperhitungkan peluang kesepakatan acak (*Chance Agreement*). **Cohen's Kappa ($\kappa$)** memberikan ukuran kesepakatan murni yang dikoreksi terhadap peluang acak ($P_o$ vs $P_e$).

## 🔬 3. Studi Kasus & Penjelasan Langkah Komputasi

Pada modul ini, kita menganalisis dataset `14_qa_bug_ratings_cohen.csv` yang mencakup data empiris pengujian sistem perangkat lunak.

**Tahapan Komputasi:**
1. *Data Ingestion & Eksplorasi*: Memuat data, memeriksa distribusi variabel, dan membersihkan anomali data.
2. *Pengujian Statistik Inferensial*: Menerapkan metode statistik parametrik/non-parametrik yang sesuai karakteristik data.
3. *Interpretasi & Rekomendasi Teknis*: Menarik kesimpulan berbasis signifikansi statistik untuk perbaikan arsitektur dan proses rekayasa software.


```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score, confusion_matrix

# 1. Ingestion Data Klasifikasi Bug oleh 2 Evaluator QA
df_qa = pd.read_csv("../datasets/14_qa_bug_ratings_cohen.csv")
display(df_qa.head())

# 2. Menghitung Matriks Kontingensi & Skor Cohen's Kappa
r1 = df_qa['rater_1_severity']
r2 = df_qa['rater_2_severity']

kappa_val = cohen_kappa_score(r1, r2)
print(f"=== Hasil Uji Kesepakatan QA ===")
print(f"Indeks Cohen's Kappa: {kappa_val:.4f}")
print("Interpretasi (Landis & Koch):", "Substantial / Almost Perfect Agreement" if kappa_val >= 0.60 else "Moderate Agreement")
```

## 📈 4. Visualisasi Grafik & Interpretasi Statistik


```python
# Visualisasi Matriks Kesepakatan Klasifikasi Bug
labels = ["Minor", "Major", "Critical"]
cm = confusion_matrix(r1, r2, labels=labels)

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Teal', xticklabels=labels, yticklabels=labels)
plt.title("Matriks Kesepakatan Klasifikasi Bug (Rater 1 vs Rater 2)", fontsize=13, fontweight='bold')
plt.xlabel("Evaluator QA 2")
plt.ylabel("Evaluator QA 1")
plt.tight_layout()
plt.show()
```

## 📝 5. Kesimpulan Saintifik & Pengambilan Keputusan

### 🎯 Kesimpulan & Actionable Insights
